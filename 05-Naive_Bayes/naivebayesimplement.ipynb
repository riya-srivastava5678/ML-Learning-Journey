```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
df_categorical_one_hot=['person_gender','person_education','person_home_ownership','loan_intent']
df_categorical_label=['previous_loan_defaults_on_file']
df = pd.get_dummies(
    df,
    columns=df_categorical_one_hot,
    dtype=int
)
le=LabelEncoder()
for col in df_categorical_label:
    df[col] = le.fit_transform(df[col])
```

```python
df
```

```python
df.columns
df=df.astype(int)
```

```python
X=df.drop('loan_status',axis=1)
y=df['loan_status']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.naive_bayes import BernoulliNB,GaussianNB
model1=GaussianNB()
model1.fit(X_train,y_train)
```

```python
y_pred=model1.predict(X_test)
```

```python
from sklearn.metrics import  accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_test,y_pred))
m=confusion_matrix(y_test,y_pred)
print(classification_report(y_test,y_pred))
```

```python
model2=BernoulliNB()
model2.fit(X_train,y_train)
```

```python
y_pred2=model2.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred2))
print(classification_report(y_test,y_pred2))
```

```python
numerical_cols = [
    "person_age",
    "person_income",
    "person_emp_exp",
    "loan_amnt",
    "loan_int_rate",
    "loan_percent_income",
    "cb_person_cred_hist_length",
    "credit_score"
]
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df = df[
        (df[col] >= lower_bound) &
        (df[col] <= upper_bound)
    ]
```

```python
df.shape
```

```python
X2=df.drop('loan_status',axis=1)
y2=df['loan_status']
```

```python
X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size=0.33, random_state=42)
```

```python
model2.fit(X_train2,y_train2)
```

```python
y3_pred=model2.predict(X_test2)

```

```python
print(accuracy_score(y_test2,y3_pred))
print(classification_report(y_test2,y3_pred))
```
