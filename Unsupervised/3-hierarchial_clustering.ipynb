```python
import numpy as np
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt
```

```python
X,y=make_blobs(
    n_samples=200,centers=3,cluster_std=2,random_state=42
)
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
single_model=AgglomerativeClustering(n_clusters=3,linkage='single')
```

```python
single_label=single_model.fit_predict(X)
```

```python
plt.scatter(X[:,0],X[:,1],c=single_label)
```

```python
complete_model=AgglomerativeClustering(n_clusters=3,linkage='complete')
complete_lables=complete_model.fit_predict(X)
```

```python
plt.scatter(X[:,0],X[:,1],c=complete_lables)
```

```python
avg_model=AgglomerativeClustering(n_clusters=3,linkage='average')
avg_labels=avg_model.fit_predict(X)
```

```python
plt.scatter(X[:,0],X[:,1],c=avg_labels)
```

```python
ward_model=AgglomerativeClustering(n_clusters=3,linkage='ward')
ward_label=ward_model.fit_predict(X)
```

```python
plt.scatter(X[:,0],X[:,1],c=ward_label)
```

```python
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
ward_linkage=linkage(X,method='ward')
```

```python
dendrogram(ward_linkage)
```
